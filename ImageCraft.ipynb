```python
#Importing Libraries
from PIL import Image
from rembg import remove
import os
```

```python
#main menu of image craft
print("===== IMAGECRAFT =====")
print("1. Resize Image")
print("2. Compress Image")
print("3. Convert Image Format")
print("4. Remove Background")
print("5. Exit")
```

```python
# Function to resize an image
def resize_image():

    # Ask the user to enter the location of the image
    path = input("Enter image path: ").strip().strip('"')

    try:
        # Open the selected image
        image = Image.open(path)

        # Ask for the new width and height
        width = int(input("Enter new width: "))
        height = int(input("Enter new height: "))

        # Check that the dimensions are positive
        if width <= 0 or height <= 0:
            print("Dimensions must be positive.")
            return

        # Resize the image to the requested dimensions
        resized = image.resize((width, height))

        # Save the resized image in PNG format
        resized.save("resized.png")

        # Display a success message
        print("Image resized successfully!")

    except Exception as error:
        # Display an error if the operation fails
        print("Error:", error)
```

```python
# Function to compress an image
def compress_image():

    # Ask the user for the image location
    path = input("Enter image path: ").strip().strip('"')

    try:
        # Open the image and convert it to RGB
        # RGB is required for standard JPEG output
        image = Image.open(path).convert("RGB")

        # Ask the user to choose image quality
        quality = int(input("Enter quality (10-95): "))

        # Validate the quality value
        if quality < 10 or quality > 95:
            print("Choose quality between 10 and 95.")
            return

        # Save the image as JPEG with compression
        # Lower quality generally creates a smaller file
        image.save(
            "compressed.jpg",
            "JPEG",
            quality=quality,
            optimize=True
        )

        print("Image compressed successfully!")

    except Exception as error:
        print("Error:", error)
```

```python
# Function to convert an image to another format
def convert_format():

    # Ask the user to enter the image location
    path = input("Enter image path: ").strip().strip('"')

    try:
        # Open the original image
        image = Image.open(path)

        # Display the available output formats
        print("1. PNG")
        print("2. JPEG")
        print("3. WEBP")

        # Ask the user to select a format
        choice = input("Choose format: ")

        # Save the image in the selected format
        if choice == "1":
            image.save("converted.png", "PNG")

        elif choice == "2":
            # JPEG does not support transparency
            image.convert("RGB").save("converted.jpg", "JPEG")

        elif choice == "3":
            image.save("converted.webp", "WEBP")

        else:
            # Handle an invalid menu selection
            print("Invalid format choice.")
            return

        print("Image converted successfully!")

    except Exception as error:
        print("Error:", error)
```

```python
# Function to remove an image background
def remove_background():

    # Ask for the image location
    path = input("Enter image path: ").strip().strip('"')

    try:
        # Open the image in binary mode
        # Binary mode reads the actual image bytes
        with open(path, "rb") as file:
            image_data = file.read()

        # Send the image data to the background-removal model
        result = remove(image_data)

        # Open an output file in binary write mode
        with open("background_removed.png", "wb") as file:

            # Write the processed image data
            file.write(result)

        print("Background removed successfully!")

    except Exception as error:
        print("Error:", error)
```

```python
#### Repeat the menu until the user chooses Exit
while True:

    # Ask the user which feature to run
    choice = input("\nEnter your choice (1-5): ")

    # Call the appropriate function
    if choice == "1":
        resize_image()

    elif choice == "2":
        compress_image()

    elif choice == "3":
        convert_format()

    elif choice == "4":
        remove_background()

    elif choice == "5":
        # Stop the program
        print("Thank you for using ImageCraft!")
        break

    else:
        # Handle an invalid menu choice
        print("Invalid choice. Try again.")
```
