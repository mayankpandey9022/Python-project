```python
#libraries ko import karne ke liye
from PIL import Image
from rembg import remove
import os
```

```python
#Project ke main menu k liye
print("===== IMAGECRAFT =====")
print("1. Resize Image")
print("2. Compress Image")
print("3. Convert Image Format")
print("4. Remove Background")
print("5. Exit")
```

```python
# image ko resize karne ke liye function
def resize_image():
    # image Path dalne ke liye
    path = input("Enter image path: ").strip().strip('"')
    try:
        # image ko open karane k liye
        image = Image.open(path)
        #width aur height user ke according dene ke liye
        width = int(input("Enter new width: "))
        height = int(input("Enter new height: "))
        #image ki dimension positive h ki nhi check krne k liye
        if width <= 0 or height <= 0:
            print("Dimensions must be positive.")
            return
        #user k according dimension dene k liye
        resized = image.resize((width, height))

        #image ko png format me save karne k liye
        resized.save("resized.png")

        # success message display karne ke liye
        print("Image resized successfully!")

    except Exception as error:
        # agar koi error aata h toh uske liye error message ko print karne ke liye
        print("Error:", error)
```

```python
# image ko compress karne ke liye function
def compress_image():

    # image ki path dene ke liye
    path = input("Enter image path: ").strip().strip('"')

    try:
        # image ko open karne ke liye aur usko rgb format dene k liye
        image = Image.open(path).convert("RGB")

        # user k according quality ke liye
        quality = int(input("Enter quality (10-95): "))

        # Quality ko validate karne ke liye
        if quality < 10 or quality > 95:
            print("Choose quality between 10 and 95.")
            return

        # image ko jgp format me convert karne ke liye
        image.save(
            "compressed.jpg",
            "JPEG",
            quality=quality,
            optimize=True
        )

        print("Image compressed successfully!")

    except Exception as error:
        print("Error:", error)
```

```python
# image ka format convert karne k liye function
def convert_format():

    # image ki path dalne ke liye
    path = input("Enter image path: ").strip().strip('"')

    try:
        # image ko open karne k liye
        image = Image.open(path)

        # formats provide karne k liye
        print("1. PNG")
        print("2. JPEG")
        print("3. WEBP")

        # user ko uske according format choose karne ke liye
        choice = input("Choose format: ")

        # user ki choice ke according image ko format me convert karne ke liye
        if choice == "1":
            image.save("converted.png", "PNG")

        elif choice == "2":
            image.convert("RGB").save("converted.jpg", "JPEG")

        elif choice == "3":
            image.save("converted.webp", "WEBP")

        else:
            # Invalid format message dispaly karne ke liye agar format invalid h toh
            print("Invalid format choice.")
            return

        print("Image converted successfully!")

    except Exception as error:
        print("Error:", error)
```

```python
# image ka background remove karne ke liye
def remove_background():

    # image ki path dalne ke liye
    path = input("Enter image path: ").strip().strip('"')

    try:
        # image ko binary mode me open karne ke liye
        with open(path, "rb") as file:
            image_data = file.read()

        # image ka background remove karne k liye
        result = remove(image_data)

        # binary file ko open karne k liye
        with open("background_removed.png", "wb") as file:

            # iamge ke process show karne k liye
            file.write(result)

        print("Background removed successfully!")

    except Exception as error:
        print("Error:", error)
```

```python
while True:

    # konsa feature user use karna chahta 
    choice = input("\nEnter your choice (1-5): ")

    # function ko call karne k liye
    if choice == "1":
        resize_image()

    elif choice == "2":
        compress_image()

    elif choice == "3":
        convert_format()

    elif choice == "4":
        remove_background()

    elif choice == "5":
        print("Thank you for using ImageCraft!")
        break

    else:
        print("Invalid choice. Try again.")
```
